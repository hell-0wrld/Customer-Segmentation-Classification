```python
import pandas as pd

# 1. 파일 읽기

df1 = pd.read_parquet('/content/drive/MyDrive/11_파이널프로젝트관련/open/train/1.회원정보/ID_Segment.parquet')
df2 = pd.read_parquet('/content/2018_train_마케팅정보.parquet')

# 2. ID 기준으로 병합
merged_df = pd.merge(df1, df2, on='ID', how='inner')  # 또는 how='outer', 'left', 'right'

print(merged_df.shape)           # (약 2400000, 약 106) 예상
print(len(merged_df.columns))    # 약 106개 예상
print(merged_df.columns[:10])

merged_df.to_parquet('마케팅정보_세그먼트병합.parquet', index=False)
print("병합되었습니다 !")
```
